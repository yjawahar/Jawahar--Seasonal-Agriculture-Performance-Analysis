```python
# %% [markdown]
# # Seasonal Agriculture Performance Analysis
#
# Run each `# %%` section as a separate Google Colab cell. Upload the supplied
# CSV when prompted. This is observational analysis: differences and
# relationships found here do not establish causation.

# %% Imports and Colab upload
!pip -q install scipy

import io
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from google.colab import files

warnings.filterwarnings("ignore", category=FutureWarning)
sns.set_theme(style="whitegrid", context="notebook")
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
RANDOM_STATE = 42

uploaded = files.upload()  # Upload seasonal_agriculture_performance_dataset.csv
csv_files = [name for name in uploaded if name.lower().endswith(".csv")]
if len(csv_files) != 1:
    raise ValueError("Please upload exactly one CSV file.")

file_path = csv_files[0]
df_raw = pd.read_csv(io.BytesIO(uploaded[file_path]))
print(f"Dataset loaded successfully: {file_path}")
print(f"Rows: {df_raw.shape[0]:,} | Columns: {df_raw.shape[1]}")

# %% Initial data understanding
display(df_raw.head())
display(df_raw.tail())
display(df_raw.sample(5, random_state=RANDOM_STATE))
print("Column names:\n", df_raw.columns.tolist())
df_raw.info()
display(df_raw.describe(include="all").T)

# %% Data quality, missing values, duplicates, and cleaning
# Median imputation is used only for numeric missing values because it is robust
# to extreme values. Categorical values would use the most common category only
# if missing; this dataset has no categorical missing values.
missing_summary = pd.DataFrame({
    "Missing_Count": df_raw.isna().sum(),
    "Missing_Percentage": (df_raw.isna().mean() * 100).round(2),
}).sort_values("Missing_Count", ascending=False)
display(missing_summary[missing_summary["Missing_Count"] > 0])

missing_plot = missing_summary.query("Missing_Count > 0").reset_index()
missing_plot = missing_plot.rename(columns={"index": "Column"})
plt.figure(figsize=(10, 4))
if missing_plot.empty:
    plt.text(.5, .5, "No missing values found", ha="center", va="center", fontsize=14)
    plt.axis("off")
else:
    sns.barplot(data=missing_plot, x="Column", y="Missing_Count", color="#4C78A8")
    plt.xticks(rotation=35, ha="right")
    plt.title("Missing values by column")
plt.tight_layout()
plt.show()

duplicate_count = df_raw.duplicated().sum()
print(f"Duplicate rows before cleaning: {duplicate_count}")
df = df_raw.drop_duplicates().copy().reset_index(drop=True)

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()
for col in numeric_columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())
for col in categorical_columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].mode(dropna=True).iat[0])

print(f"Shape after cleaning: {df.shape}")
print(f"Remaining missing values: {df.isna().sum().sum()}")

# %% Variable review
# Identifier: Farm_ID
# Categorical/location: State, District, Crop, Season, Irrigation_Method
# Environmental: rainfall, temperature, humidity, sunlight, soil pH/moisture
# Operational: area, nutrients, fertilizer, pesticide, seed quality, water use
# Performance/economic: yield, production, price, costs, revenue, profit, efficiency
print("Numerical columns:\n", numeric_columns)
print("\nCategorical columns:\n", categorical_columns)
for col in [c for c in categorical_columns if c != "Farm_ID"]:
    print(f"{col} ({df[col].nunique()} levels): {sorted(df[col].dropna().unique().tolist())}")

# %% Descriptive statistics and seasonal comparison
statistical_summary = df[numeric_columns].describe().T
statistical_summary["Range"] = statistical_summary["max"] - statistical_summary["min"]
statistical_summary["IQR"] = statistical_summary["75%"] - statistical_summary["25%"]
display(statistical_summary.round(2))

season_order = [s for s in ["Kharif", "Rabi", "Zaid"] if s in df["Season"].unique()]
season_order += [s for s in df["Season"].unique() if s not in season_order]
df["Profit_Margin_pct"] = np.where(
    df["Revenue_INR"].ne(0), 100 * df["Profit_INR"] / df["Revenue_INR"], np.nan
)
season_metrics = df.groupby("Season", observed=True).agg(
    Records=("Farm_ID", "size"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Profit=("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Profit_Margin_pct=("Profit_Margin_pct", "mean"),
).reindex(season_order).round(2)
display(season_metrics)

for outcome in ["Yield_Tonnes_Ha", "Profit_INR", "Water_Used_m3"]:
    groups = [group[outcome].dropna() for _, group in df.groupby("Season", observed=True)]
    h_stat, p_value = stats.kruskal(*groups)
    print(f"Kruskal-Wallis for {outcome}: H={h_stat:.2f}, p={p_value:.4g}")
print("Small p-values indicate different observed distributions, not a cause of the difference.")

# %% Univariate analysis
fig, axes = plt.subplots(2, 3, figsize=(17, 9))
sns.countplot(data=df, x="Season", order=season_order, ax=axes[0, 0], color="#59A14F")
axes[0, 0].set_title("Records by season")
sns.countplot(data=df, x="Crop", order=df["Crop"].value_counts().index, ax=axes[0, 1], color="#F28E2B")
axes[0, 1].tick_params(axis="x", rotation=35); axes[0, 1].set_title("Records by crop")
for ax, column, title in [
    (axes[0, 2], "Yield_Tonnes_Ha", "Yield distribution"),
    (axes[1, 0], "Profit_INR", "Profit distribution"),
    (axes[1, 1], "Rainfall_mm", "Rainfall distribution"),
]:
    sns.histplot(data=df, x=column, kde=True, ax=ax, color="#4C78A8")
    ax.set_title(title)
sns.boxplot(data=df, y="Yield_Tonnes_Ha", ax=axes[1, 2], color="#B07AA1")
axes[1, 2].set_title("Yield: spread and potential outliers")
plt.tight_layout(); plt.show()

# %% IQR outlier review - do not delete flagged observations automatically
outlier_rows = []
for col in numeric_columns:
    q1, q3 = df[col].quantile([.25, .75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_rows.append([col, lower, upper, count, 100 * count / len(df)])
outlier_summary = pd.DataFrame(
    outlier_rows,
    columns=["Column", "Lower_Bound", "Upper_Bound", "Outlier_Count", "Outlier_Percentage"],
).sort_values("Outlier_Count", ascending=False)
display(outlier_summary.round(2))

# %% Bivariate analysis
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
sns.boxplot(data=df, x="Season", y="Yield_Tonnes_Ha", order=season_order, ax=axes[0, 0])
axes[0, 0].set_title("Yield across seasons")
sns.boxplot(data=df, x="Season", y="Profit_INR", order=season_order, ax=axes[0, 1])
axes[0, 1].set_title("Profit across seasons")
sns.barplot(data=df, x="Season", y="Water_Used_m3", order=season_order, errorbar=("ci", 95), ax=axes[0, 2])
axes[0, 2].set_title("Mean water use (95% CI)")
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", alpha=.35, ax=axes[1, 0])
axes[1, 0].set_title("Rainfall versus yield")
sns.scatterplot(data=df, x="Farm_Area_Hectares", y="Production_Tonnes", alpha=.35, ax=axes[1, 1])
axes[1, 1].set_title("Farm area versus production")
sns.regplot(data=df, x="Yield_Tonnes_Ha", y="Profit_INR", scatter_kws={"alpha": .2}, line_kws={"color": "black"}, ax=axes[1, 2])
axes[1, 2].set_title("Yield versus profit")
plt.tight_layout(); plt.show()

# %% Multivariate analysis and correlation
fig, axes = plt.subplots(1, 2, figsize=(19, 6))
sns.boxplot(data=df, x="Crop", y="Yield_Tonnes_Ha", hue="Season", hue_order=season_order, ax=axes[0])
axes[0].tick_params(axis="x", rotation=35); axes[0].set_title("Yield by crop and season")
sns.scatterplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", hue="Season", hue_order=season_order,
                size="Farm_Area_Hectares", sizes=(15, 160), alpha=.55, ax=axes[1])
axes[1].set_title("Rainfall, yield, farm area and season")
plt.tight_layout(); plt.show()

plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Irrigation_Method", y="Yield_Tonnes_Ha", hue="Season", hue_order=season_order)
plt.title("Yield by irrigation method and season")
plt.tight_layout(); plt.show()

plt.figure(figsize=(16, 12))
sns.heatmap(df[numeric_columns].corr(), cmap="coolwarm", center=0, annot=True, fmt=".2f", linewidths=.3)
plt.title("Correlation matrix of numerical variables")
plt.tight_layout(); plt.show()

# %% Three additional student-designed analyses
# A. Crop-specific season results test whether overall seasonal patterns persist by crop.
crop_season_summary = df.groupby(["Season", "Crop"], observed=True).agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Records=("Farm_ID", "size"),
).round(2).sort_values(["Season", "Average_Yield"], ascending=[True, False])
display(crop_season_summary)

# B. Loss rate and profit margin address risk that the mean profit can conceal.
profit_risk = df.assign(Loss=df["Profit_INR"] < 0).groupby("Season", observed=True).agg(
    Loss_Rate_pct=("Loss", lambda values: 100 * values.mean()),
    Median_Profit=("Profit_INR", "median"),
    Mean_Profit_Margin_pct=("Profit_Margin_pct", "mean"),
).reindex(season_order).round(2)
display(profit_risk)

# C. Irrigation comparison evaluates water use, efficiency, and yield jointly.
irrigation_summary = df.groupby(["Season", "Irrigation_Method"], observed=True).agg(
    Records=("Farm_ID", "size"), Mean_Water_Used=("Water_Used_m3", "mean"),
    Mean_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Mean_Yield=("Yield_Tonnes_Ha", "mean"),
).round(2)
display(irrigation_summary)

# D. Regional variation identifies state-season combinations needing follow-up.
regional_summary = df.groupby(["State", "Season"], observed=True).agg(
    Records=("Farm_ID", "size"), Mean_Yield=("Yield_Tonnes_Ha", "mean"),
    Mean_Profit=("Profit_INR", "mean"),
).round(2).sort_values("Mean_Profit")
display(regional_summary)

# %% Reproducible written findings, recommendations, and conclusion
best_yield_season = season_metrics["Average_Yield"].idxmax()
best_profit_season = season_metrics["Average_Profit"].idxmax()
highest_loss_season = profit_risk["Loss_Rate_pct"].idxmax()
highest_water_season = season_metrics["Average_Water_Used"].idxmax()
yield_p = stats.kruskal(*[group["Yield_Tonnes_Ha"] for _, group in df.groupby("Season", observed=True)]).pvalue
profit_p = stats.kruskal(*[group["Profit_INR"] for _, group in df.groupby("Season", observed=True)]).pvalue

insights = [
    f"{best_yield_season} has the highest average yield ({season_metrics.loc[best_yield_season, 'Average_Yield']:.2f} tonnes/ha). Evidence: seasonal table and boxplot. Limitation: crop mix and inputs may differ by season.",
    f"{best_profit_season} has the highest average profit (INR {season_metrics.loc[best_profit_season, 'Average_Profit']:,.0f}). Evidence: seasonal table. Limitation: averages can hide farm-level variation.",
    f"{highest_loss_season} has the highest loss rate ({profit_risk.loc[highest_loss_season, 'Loss_Rate_pct']:.1f}%). Evidence: profitability-risk table. Limitation: this does not prove season caused the losses.",
    f"{highest_water_season} has the largest average water use ({season_metrics.loc[highest_water_season, 'Average_Water_Used']:,.0f} m3). Evidence: seasonal water comparison. Limitation: area and irrigation mix may confound it.",
    f"Yield distributions differ across seasons (Kruskal-Wallis p={yield_p:.4g}). Evidence: test and boxplot. Limitation: statistical significance is not practical importance.",
    f"Profit distributions differ across seasons (Kruskal-Wallis p={profit_p:.4g}). Evidence: test and boxplot. Limitation: the test does not explain the difference.",
    "Crop-season results show whether the overall pattern remains within crop groups. Evidence: crop-season table. Limitation: small groups require caution.",
    "Irrigation-season results compare water use, efficiency, and yield together. Evidence: irrigation table and boxplot. Limitation: irrigation was not randomly assigned.",
]
print("KEY INSIGHTS\n")
print("\n\n".join(f"{i + 1}. {text}" for i, text in enumerate(insights)))

print("\nRECOMMENDATIONS")
print(f"- Prioritize diagnostic follow-up for {highest_loss_season}, which has the highest observed loss rate; inspect crop, state, costs, and irrigation before intervention.")
print(f"- Compare practices in {best_yield_season} with other seasons within the same crop and state; avoid direct cross-season assumptions.")
print(f"- Review water management in {highest_water_season}, concentrating on irrigation groups with low observed water efficiency.")
print("- Validate IQR-flagged records against source systems before excluding them.")

print("\nCONCLUSION")
print("The dataset shows measurable seasonal differences in yield, profit, and resource use. It supports targeted crop-, state-, and irrigation-specific follow-up, but its observational design cannot establish causal effects.")
# Multivariate: Pair Plot of Key Environmental and Performance Metrics by Season

selected_columns = ['Avg_Temperature_C', 'Rainfall_mm', 'Humidity_pct', 'Yield_Tonnes_Ha', 'Profit_INR', 'Season']
sns.pairplot(df[selected_columns], hue='Season', palette='viridis', plot_kws={'alpha': 0.6})
plt.suptitle('Pair Plot of Environmental and Performance Metrics by Season', y=1.02) # Adjust suptitle position
plt.tight_layout()
plt.show()
```
